```python
import json, os, shutil, tarfile
from pathlib import Path
from google.colab import drive

if not os.path.ismount("/content/drive"):
    drive.mount("/content/drive")
assert os.path.ismount("/content/drive"), "Drive is NOT mounted!"

SEG = Path("/content/drive/MyDrive/ai-medical-imaging/segmentation")
OUT = SEG / "ct_demo"
OUT.mkdir(exist_ok=True)

split = json.load(open(SEG / "results" / "split.json"))
picks = [split["test"][0], split["test"][2]]   # Dice 0.82 (weakest) and Dice 0.96

with tarfile.open(SEG / "Task09_Spleen.tar") as tar:
    for item in picks:
        name = Path(item["image"]).name
        for folder, out_name in [("imagesTr", name), ("labelsTr", "label_" + name)]:
            f = tar.extractfile(tar.getmember(f"Task09_Spleen/{folder}/{name}"))
            with open(OUT / out_name, "wb") as g:
                shutil.copyfileobj(f, g)
            print("saved", out_name, round((OUT / out_name).stat().st_size / 1e6, 1), "MB")
```
